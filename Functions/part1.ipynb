```python
import pandas as pd
def operations_grade(marks):
    if marks <=20:
        return "A"
    elif marks >=40 or marks <= 80:
        return "B"
data_dict={
    'marks':[40,50,40,50,67,90]
}
df1=pd.DataFrame(data=data_dict,columns=['marks'])
results=df1['marks'].apply(operations_grade)
print(df1['marks'],results)
```
