```python
x=df2[['studyName','Sample Number','Region']]
y=df2['Species_numeric']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,test_size=0.2,random_state=42)
```

```python
NGassian=GaussianNB()
NGassian.fit(x_train,y_train)
```

```python
NG_predict=NGassian.predict(x_test)
print(NG_predict)
```

```python
New_Samples=NGassian.predict([[1,1,3]])
print(New_Samples)
```

```python
NG_Predict_pro=NGassian.predict_proba(x_test)
print(NG_Predict_pro)
```

```python
print(accuracy_score(y_test,NG_predict))
```
