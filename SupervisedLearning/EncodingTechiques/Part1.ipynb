```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns 
from sklearn.preprocessing import LabelEncoder,OneHotEncoder,OrdinalEncoder
from category_encoders import TargetEncoder,CountEncoder,BinaryEncoder,SummaryEncoder,SumEncoder,GrayEncoder
```

```python
data = {
    'Color': ['Red', 'Blue', 'Green', 'Blue', 'Red', 'Green', 'Blue', 'Red'],
    'Size': ['S', 'M', 'L', 'M', 'S', 'XL', 'L', 'M'],
    'City': ['New York', 'London', 'Paris', 'London', 'New York', 'Tokyo', 'Paris', 'New York'],
    'Purchased': [1, 0, 1, 0, 1, 0, 1, 0] # This will be our target variable for some encoders
}
```

```python
df1=pd.DataFrame(data)

df1
```

##lable encoding used in 

nomial data 
No order varubales
returns the int values
Unique integer to each class for column
if target variable is spam/not like that such conditions like y variables

```python
label_encoding=LabelEncoder()
df1['color_label_encoded']=label_encoding.fit_transform(df1['Color'])
print(df1[['Color','color_label_encoded']])
```

##ordinal encoding
-> When order is matter
-> Specially for features 
-> returns the float value
-> customised order classes data

```python
orders=['S','M','XL','L']

ordinal_encode=OrdinalEncoder(categories=[orders],handle_unknown='use_encoded_value',unknown_value=10,encoded_missing_value=12)
df1['size_encoded']=ordinal_encode.fit_transform(df1[['Size']]).astype(int)
print(df1[['Size','size_encoded']])
```

##one hot encoder

```python
#using sklearn

cities=['New York', 'London', 'Paris', 'Tokyo']
Ohe=OneHotEncoder(categories=[cities],handle_unknown='ignore',sparse_output=False,drop='first')

cities_ohe_encoded=Ohe.fit_transform(df1[['City']])
print(cities_ohe_encoded.astype(int))
```
