```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression

# Generate linear data
np.random.seed(42)
# WHY: Create reproducible results
# WHAT: Sets random number generator seed

X_linear = np.linspace(0, 10, 100).reshape(-1, 1)
# WHY: Create feature values from 0 to 10
# WHAT: Generates 100 evenly spaced numbers and reshapes to column vector

y_linear = 2 * X_linear.ravel() + 1 + np.random.normal(0, 0.5, 100)
# WHY: Create linear relationship with some noise
# WHAT: y = 2x + 1 + noise, ravel() converts 2D to 1D

# Split the data
X_train_lin, X_test_lin, y_train_lin, y_test_lin = train_test_split(
    X_linear, y_linear, test_size=0.2, random_state=42
)
# WHY: Evaluate model on unseen data
# WHAT: 80% training, 20% testing

# Create and train Decision Tree for linear data
dt_linear = DecisionTreeRegressor(
    max_depth=3,              # WHY: Limit tree depth for linear data
    min_samples_split=20,     # WHY: Require more samples to split
    min_samples_leaf=10,      # WHY: Prevent overfitting
    random_state=42
)
# WHAT: Decision tree configured for linear patterns

dt_linear.fit(X_train_lin, y_train_lin)
# WHY: Train model on training data
# WHAT: Builds decision tree based on training data

# Compare with Linear Regression
lr = LinearRegression()
lr.fit(X_train_lin, y_train_lin)
# WHY: Compare with traditional linear model
# WHAT: Fits straight line to data

# Make predictions
y_pred_dt_linear = dt_linear.predict(X_test_lin)
y_pred_lr_linear = lr.predict(X_test_lin)
# WHY: Evaluate model performance
# WHAT: Predict target values for test features

# Calculate metrics
mse_dt = mean_squared_error(y_test_lin, y_pred_dt_linear)
mse_lr = mean_squared_error(y_test_lin, y_pred_lr_linear)
r2_dt = r2_score(y_test_lin, y_pred_dt_linear)
r2_lr = r2_score(y_test_lin, y_pred_lr_linear)
# WHY: Quantify model performance
# WHAT: MSE measures average squared difference, R² measures goodness of fit

print("Linear Data Results:")
print(f"Decision Tree - MSE: {mse_dt:.4f}, R²: {r2_dt:.4f}")
print(f"Linear Regression - MSE: {mse_lr:.4f}, R²: {r2_lr:.4f}")
# WHY: Compare performance between models

# Visualize results
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.scatter(X_linear, y_linear, alpha=0.7, label='Actual Data')
plt.plot(X_linear, lr.predict(X_linear), color='red', linewidth=2, label='Linear Regression')
plt.title('Linear Regression Fit')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()

plt.subplot(1, 3, 2)
plt.scatter(X_linear, y_linear, alpha=0.7, label='Actual Data')
plt.scatter(X_test_lin, y_pred_dt_linear, color='green', label='Decision Tree Predictions')
plt.title('Decision Tree Predictions')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()

plt.subplot(1, 3, 3)
# Visualize decision tree
plot_tree(dt_linear, filled=True, feature_names=['X'])
plt.title('Decision Tree Structure')
plt.show()
```
