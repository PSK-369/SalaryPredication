```python
# Generate non-linear data
np.random.seed(42)
X_nonlinear = np.linspace(0, 10, 100).reshape(-1, 1)
y_nonlinear = np.sin(X_nonlinear.ravel()) * 2 + np.random.normal(0, 0.3, 100)
# WHY: Create sine wave pattern with noise
# WHAT: y = 2*sin(x) + noise - non-linear relationship

# Split the data
X_train_nl, X_test_nl, y_train_nl, y_test_nl = train_test_split(
    X_nonlinear, y_nonlinear, test_size=0.2, random_state=42
)

# Create Decision Tree for non-linear data
dt_nonlinear = DecisionTreeRegressor(
    max_depth=5,              # WHY: Deeper tree to capture non-linearity
    min_samples_split=5,      # WHY: Allow more splits for complex patterns
    min_samples_leaf=3,       # WHY: Smaller leaves for detailed patterns
    random_state=42
)
# WHAT: Decision tree configured for complex, non-linear relationships

dt_nonlinear.fit(X_train_nl, y_train_nl)

# Compare with Linear Regression (will perform poorly)
lr_nl = LinearRegression()
lr_nl.fit(X_train_nl, y_train_nl)

# Make predictions
y_pred_dt_nonlinear = dt_nonlinear.predict(X_test_nl)
y_pred_lr_nonlinear = lr_nl.predict(X_test_nl)

# Calculate metrics
mse_dt_nl = mean_squared_error(y_test_nl, y_pred_dt_nonlinear)
mse_lr_nl = mean_squared_error(y_test_nl, y_pred_lr_nonlinear)
r2_dt_nl = r2_score(y_test_nl, y_pred_dt_nonlinear)
r2_lr_nl = r2_score(y_test_nl, y_pred_lr_nonlinear)

print("\nNon-linear Data Results:")
print(f"Decision Tree - MSE: {mse_dt_nl:.4f}, R²: {r2_dt_nl:.4f}")
print(f"Linear Regression - MSE: {mse_lr_nl:.4f}, R²: {r2_lr_nl:.4f}")

# Visualize comparison
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
x_plot = np.linspace(0, 10, 1000).reshape(-1, 1)
plt.scatter(X_nonlinear, y_nonlinear, alpha=0.7, label='Actual Data')
plt.plot(x_plot, lr_nl.predict(x_plot), color='red', linewidth=2, label='Linear Regression')
plt.title('Linear Regression on Non-linear Data')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()

plt.subplot(1, 3, 2)
plt.scatter(X_nonlinear, y_nonlinear, alpha=0.7, label='Actual Data')
plt.plot(x_plot, dt_nonlinear.predict(x_plot), color='green', linewidth=2, label='Decision Tree')
plt.title('Decision Tree on Non-linear Data')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()

plt.subplot(1, 3, 3)
plot_tree(dt_nonlinear, filled=True, feature_names=['X'])
plt.title('Non-linear Data Decision Tree')
plt.tight_layout()
plt.show()
```
