```python
from sklearn.datasets import make_moons, make_circles

# Generate non-linear data (moons pattern)
X_nonlinear_clf, y_nonlinear_clf = make_moons(
    n_samples=1000,
    noise=0.1,
    random_state=42
)
# WHY: Create non-linearly separable data (two interleaving half-circles)
# WHAT: Generates moon-shaped clusters that can't be separated by straight line

# Split the data
X_train_nl_clf, X_test_nl_clf, y_train_nl_clf, y_test_nl_clf = train_test_split(
    X_nonlinear_clf, y_nonlinear_clf, test_size=0.2, random_state=42
)

# Create Decision Tree for non-linear classification
dt_nonlinear_clf = DecisionTreeClassifier(
    max_depth=5,              # WHY: Deeper tree to capture complex boundaries
    min_samples_split=10,     # WHY: Allow more splits for complex patterns
    min_samples_leaf=5,       # WHY: Smaller leaves for detailed boundaries
    criterion='entropy',      # WHY: Use entropy for potentially better splits
    random_state=42
)
# WHAT: Classification tree configured for complex, non-linear decision boundaries

dt_nonlinear_clf.fit(X_train_nl_clf, y_train_nl_clf)

# Compare with Logistic Regression (will perform poorly)
logreg_nl = LogisticRegression(random_state=42)
logreg_nl.fit(X_train_nl_clf, y_train_nl_clf)

# Make predictions
y_pred_dt_nonlinear_clf = dt_nonlinear_clf.predict(X_test_nl_clf)
y_pred_logreg_nonlinear_clf = logreg_nl.predict(X_test_nl_clf)

# Calculate accuracy
acc_dt_nonlinear = accuracy_score(y_test_nl_clf, y_pred_dt_nonlinear_clf)
acc_logreg_nonlinear = accuracy_score(y_test_nl_clf, y_pred_logreg_nonlinear_clf)

print("\nNon-linear Classification Results:")
print(f"Decision Tree Accuracy: {acc_dt_nonlinear:.4f}")
print(f"Logistic Regression Accuracy: {acc_logreg_nonlinear:.4f}")

# Detailed classification report
print("\nDecision Tree Classification Report:")
print(classification_report(y_test_nl_clf, y_pred_dt_nonlinear_clf))

# Confusion Matrix
cm = confusion_matrix(y_test_nl_clf, y_pred_dt_nonlinear_clf)
print("Confusion Matrix:")
print(cm)

# Visualize non-linear decision boundaries
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.scatter(X_nonlinear_clf[:, 0], X_nonlinear_clf[:, 1], c=y_nonlinear_clf, cmap='RdYlBu')
plt.title('Original Non-linear Data (Moons)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

plt.subplot(1, 3, 2)
plot_decision_boundaries(X_nonlinear_clf, y_nonlinear_clf, logreg_nl, 'Logistic Regression on Non-linear Data')

plt.subplot(1, 3, 3)
plot_decision_boundaries(X_nonlinear_clf, y_nonlinear_clf, dt_nonlinear_clf, 'Decision Tree on Non-linear Data')

plt.tight_layout()
plt.show()

# Show complex tree structure
plt.figure(figsize=(15, 10))
plot_tree(dt_nonlinear_clf, filled=True, feature_names=['Feature 1', 'Feature 2'], 
          class_names=['Class 0', 'Class 1'], fontsize=10)
plt.title('Decision Tree for Non-linear Classification')
plt.show()

# Feature importance analysis
feature_importance = dt_nonlinear_clf.feature_importances_
print(f"\nFeature Importance: {feature_importance}")
# WHY: Understand which features contribute most to decisions
# WHAT: Shows relative importance of each feature in splits
```
