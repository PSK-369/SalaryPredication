```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.linear_model import LogisticRegression

# Generate linearly separable data
X_linear_clf, y_linear_clf = make_classification(
    n_samples=1000,
    n_features=2,
    n_redundant=0,
    n_informative=2,
    n_clusters_per_class=1,
    random_state=42
)
# WHY: Create dataset where classes can be separated by straight line
# WHAT: Generates 1000 samples with 2 features, 2 classes

# Split the data
X_train_lin_clf, X_test_lin_clf, y_train_lin_clf, y_test_lin_clf = train_test_split(
    X_linear_clf, y_linear_clf, test_size=0.2, random_state=42
)

# Create Decision Tree classifier for linear data
dt_linear_clf = DecisionTreeClassifier(
    max_depth=3,              # WHY: Shallow tree for simple linear separation
    min_samples_split=50,     # WHY: Prevent over-splitting
    min_samples_leaf=20,      # WHY: Larger leaves for generalization
    criterion='gini',         # WHY: Use Gini impurity (faster computation)
    random_state=42
)
# WHAT: Classification tree optimized for linearly separable data

dt_linear_clf.fit(X_train_lin_clf, y_train_lin_clf)
# WHY: Train classifier on training data
# WHAT: Builds decision boundaries based on feature splits

# Compare with Logistic Regression
logreg = LogisticRegression(random_state=42)
logreg.fit(X_train_lin_clf, y_train_lin_clf)
# WHY: Compare with linear classification model
# WHAT: Fits logistic function for binary classification

# Make predictions
y_pred_dt_linear_clf = dt_linear_clf.predict(X_test_lin_clf)
y_pred_logreg_linear_clf = logreg.predict(X_test_lin_clf)
# WHY: Evaluate classification performance
# WHAT: Predict class labels for test data

# Calculate accuracy
acc_dt_linear = accuracy_score(y_test_lin_clf, y_pred_dt_linear_clf)
acc_logreg_linear = accuracy_score(y_test_lin_clf, y_pred_logreg_linear_clf)
# WHY: Measure classification performance
# WHAT: Percentage of correct predictions

print("Linear Classification Results:")
print(f"Decision Tree Accuracy: {acc_dt_linear:.4f}")
print(f"Logistic Regression Accuracy: {acc_logreg_linear:.4f}")

# Visualize decision boundaries
def plot_decision_boundaries(X, y, model, title):
    # WHY: Visualize how model separates classes
    # WHAT: Creates contour plot of decision regions
    
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    # WHY: Set plot boundaries with padding
    
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                         np.arange(y_min, y_max, 0.1))
    # WHY: Create grid of points for prediction
    
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    # WHY: Predict class for each grid point
    # WHAT: ravel() flattens, c_ concatenates columns
    
    Z = Z.reshape(xx.shape)
    # WHY: Reshape predictions to match grid shape
    
    plt.contourf(xx, yy, Z, alpha=0.4, cmap='RdYlBu')
    # WHY: Fill decision regions with colors
    
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap='RdYlBu', edgecolors='k')
    # WHY: Plot actual data points
    plt.title(title)
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.scatter(X_linear_clf[:, 0], X_linear_clf[:, 1], c=y_linear_clf, cmap='RdYlBu')
plt.title('Original Linear Data')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

plt.subplot(1, 3, 2)
plot_decision_boundaries(X_linear_clf, y_linear_clf, logreg, 'Logistic Regression Boundaries')

plt.subplot(1, 3, 3)
plot_decision_boundaries(X_linear_clf, y_linear_clf, dt_linear_clf, 'Decision Tree Boundaries')

plt.tight_layout()
plt.show()

# Show tree structure
plt.figure(figsize=(12, 8))
plot_tree(dt_linear_clf, filled=True, feature_names=['Feature 1', 'Feature 2'], 
          class_names=['Class 0', 'Class 1'])
plt.title('Decision Tree for Linear Classification')
plt.show()
```
