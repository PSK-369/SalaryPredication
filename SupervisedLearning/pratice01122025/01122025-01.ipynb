```python
class Anomoilie_detection:
    def __init__(self,data):
        self.data=data
    def checking_function(self):
        anomiles=[]
        mean=sts.mean(self.data)
        std=sts.stdev(self.data)
        cutoff=3*std
        higher_bound=mean+cutoff
        lower_bound=mean-cutoff
        for each_point in self.data:
            if each_point > higher_bound or each_point < lower_bound:
                anomiles.append(each_point)
        return anomiles
    
anomiles_instance=Anomoilie_detection(dataframe3['Years'])
my_anomiles_data=anomiles_instance.checking_function()
print(my_anomiles_data)
```

```python
import numpy as np
from abc import ABC, abstractmethod

class List_class(ABC):
    @abstractmethod
    def append_function1(self, item):
        pass

    @abstractmethod
    def extend_function(self, items):
        pass

    @abstractmethod
    def insert_function(self, index, item):
        pass

    @abstractmethod
    def delete_function(self, index):
        pass

    @abstractmethod
    def clear_function(self):
        pass

    @abstractmethod
    def pop_function(self, index=-1):
        pass

    @abstractmethod
    def remove_function(self, item):
        pass

    @abstractmethod
    def sort_function(self):
        pass

    @abstractmethod
    def temporary_sorted(self):
        pass

    @abstractmethod
    def reverse_sort(self):
        pass

class Implementation_abstraction(List_class):
    def __init__(self, data):
        self.data = list(data)  # make a copy to avoid side effects

    def append_function1(self, item):
        self.data.append(item)
        return self.data

    def extend_function(self, items):
        self.data.extend(items)
        return self.data

    def insert_function(self, index, item):
        self.data.insert(index, item)
        return self.data

    def delete_function(self, index):
        del self.data[index]
        return self.data

    def clear_function(self):
        self.data.clear()
        return self.data  # now returns [] (empty list)

    def pop_function(self, index=-1):
        return self.data.pop(index)  # returns popped value

    def remove_function(self, item):
        self.data.remove(item)
        return self.data

    def sort_function(self):
        self.data.sort()
        return self.data

    def temporary_sorted(self):
        return sorted(self.data)  # returns new sorted list, doesn't modify self.data

    def reverse_sort(self):
        self.data.sort(reverse=True)
        return self.data

# Usage
list1 = [1, 20, 3, 4, 5, 6, 7, 8, 10]
impl = Implementation_abstraction(list1)

print("Original:", impl.data)
print("Sorted:", impl.sort_function())
print("Temporary sorted (original unchanged after this):", impl.temporary_sorted())
print("After sort:", impl.data)
```
