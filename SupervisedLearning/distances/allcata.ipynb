```python
from scipy.spatial import distance

point1=(1,2,3)
point2=(4,5,6)

newEclidian=distance.euclidean(point1,point2)
print(newEclidian)
```

```python
newManha=distance.cityblock(point1,point2)
print(newManha)
```

```python
newMoniki=distance.minkowski(point1,point2,p=2)
print(newMoniki)
```

```python
newCosine=distance.cosine(point1,point2)
print(newCosine)
```

```python
newSupermum=distance.hamming(point1,point2)
print(newSupermum)
```
