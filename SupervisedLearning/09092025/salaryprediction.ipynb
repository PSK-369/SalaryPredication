```python

np.random.seed(42)
x_value=np.random.randint(low=1,high=20,size=(100))
y_value=np.random.randint(low=100000,high=1000000,size=(100))
```

```python
def anomiles_detection(y):
    ''' hey shravan anomiles detection function series values'''
    anomolies=[]
    std_devi=sts.stdev(y)
    avg=sts.mean(y)
    normal_standarddeviation=3*sts.stdev(y)
    left_cuffoff=sts.mean(y)-normal_standarddeviation
    right_cutoff=sts.mean(y)+normal_standarddeviation
    
    print(f" standard devi: {std_devi}\n avg: {avg}\n normal_distribution is : {normal_standarddeviation}\n left cuttoff is : {left_cuffoff}\n right cuttoff is : {right_cutoff}")
    for each_anomiles_point in y:
        if each_anomiles_point > right_cutoff or each_anomiles_point < left_cuffoff:
           anomolies.append(each_anomiles_point)    
    return anomolies
y_values1=[1,2,3,4,5,6,90,19,100,212,900]
z=anomiles_detection(y_values1)
print('\n')
if np.empty(z):
    print('empty list because no anomiloes')
else:
    print(f"{z}")
```

```python
x=df1['experince'].values.reshape(-1,1)
y=df1['salary']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,train_size=0.8,random_state=42)
```

```python
regreesionmodel=LinearRegression()
regreesionmodel.fit(x_train,y_train)
```

```python
y_pred=regreesionmodel.predict(x_test)
print(y_pred)
```

```python
print(regreesionmodel.intercept_)
print(regreesionmodel.coef_[0])
print(abs(r2_score(y_test,y_pred)))
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
print(median_absolute_error(y_test,y_pred))
print(root_mean_squared_error(y_test,y_pred))
```
