```python
import numpy as np 
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.preprocessing import (StandardScaler,MinMaxScaler,PolynomialFeatures)
from sklearn.metrics import (mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score,mean_absolute_percentage_error,median_absolute_error)
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.linear_model import LinearRegression
import seaborn as sns
```

```python
np.random.seed(42)
x=np.random.rand(100,2)
y=np.random.rand(100,2)
z=10*x*y+np.random.rand(100,2).round(2)
```

```python
x_test,x_train,y_test,y_train=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model =LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(y_test)
```

```python
#performcase metrics
mae=mean_absolute_error(y_test,y_pred)

print(f"mean abs error {mae}")

mse=mean_squared_error(y_test,y_pred)
print(f"mean square err {mse}")

r2score=r2_score(y_test,y_pred)
print(f"r2score {abs(r2score)}")

print(f"score is {model.score(y_test,y_pred)}")

print(f"mean sq percenate error {mean_absolute_percentage_error(y_test,y_pred)}")
print(f"root_mean_squared_error {root_mean_squared_error(y_test,y_pred)}")
print(model.intercept_[1])

print(model.coef_)
```
