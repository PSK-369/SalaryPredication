```python
import pandas as pd
from scipy.stats import spearmanr

df = pd.DataFrame({
    'Height': [150, 160, 170, 180, 190],
    'Weight': [50, 60, 65, 180, 90]
})

# Spearman correlation matrix (entire DataFrame)
spearman_matrix = df.corr(method='spearman')
print(spearman_matrix)

# Or use scipy for coefficient + p-value
rho, p = spearmanr(df['Height'], df['Weight'])
print(f"ρ = {rho:.3f}, p = {p:.3f}")
```
