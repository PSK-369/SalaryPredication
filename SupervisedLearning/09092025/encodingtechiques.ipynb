```python
np.random.seed(42)
x=np.random.randint(low=1,high=100,size=1000).reshape(-1,1)
y=5*x+np.random.randint(low=1,high=100,size=1000)
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,train_size=0.8,random_state=42)
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
print(f"{y_pred}")
```

```python
max_scaler=MinMaxScaler()
result_max_min_scaler=max_scaler.fit_transform(X_test,y_pred)
print(result_max_min_scaler)

```

```python
colors=['#ff8c00']
sns.set_theme(style='whitegrid')
sns.set_palette(colors)
sns.kdeplot(result_max_min_scaler)

plt.show()
```
