```python
import numpy as np 
import pandas as pd 
from matplotlib import pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler,MinMaxScaler,PolynomialFeatures
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.pipeline import Pipeline
```

```python
np.random.seed(42)
x=np.random.randint(low=1,high=200,size=(200)).reshape(-1,1)
y=np.random.randint(low=1,high=200,size=200)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
Linear_regression=LinearRegression()

cross_score=cross_val_score(Linear_regression,x,y,scoring='neg_mean_squared_error',cv=10)
mse_error=np.mean(cross_score)
print(abs(mse_error))


poly_features_3=PolynomialFeatures(degree=3,include_bias=True)
poly_features_4=PolynomialFeatures(degree=9,include_bias=True)
models_poly=Pipeline([
('poly',poly_features_3),
('linear',Linear_regression)    
])
models_poly1=Pipeline([
('poly1',poly_features_4),
('linear',Linear_regression)    
])

models={
    'linear':Linear_regression,
    'models_poly1':models_poly,
    'models_poly2':models_poly1
}


for name,model in models.items():
    print("name is ",name)
    model.fit(x_train,y_train)
    y_pred1=model.predict(x_test)
    print(y_pred1)
    print(model.score(x_test,y_pred1))
```
