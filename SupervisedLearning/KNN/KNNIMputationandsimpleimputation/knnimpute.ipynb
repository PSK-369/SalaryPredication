```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

# Data
dict_data = {
   "stu_marks": [89,29,36,45,50,56,90,23,412,523,25,29,np.nan,90,67,56,np.nan,90,np.nan],
   'stu_final_marks': [890,294,3,5,50,6,np.nan,23,412,523,25,29,np.nan,90,67,56,np.nan,90,np.nan]   
}

df1 = pd.DataFrame(dict_data)
# # Optional: Better imputation with scaling
# scaler = StandardScaler()
# data_scaled = scaler.fit_transform(df1[['stu_marks', 'stu_final_marks']])
# data_imputed_scaled = knn_imputer.fit_transform(data_scaled)
# data_imputed = scaler.inverse_transform(data_imputed_scaled)
# df1[['stu_marks1', 'stu_final_marks1']] = data_imputed.astype(int)

# Impute missing values
knn_imputer = KNNImputer(n_neighbors=6, weights='distance', metric='nan_euclidean')
df1[['stu_marks1', 'stu_final_marks1']] = knn_imputer.fit_transform(df1[['stu_marks', 'stu_final_marks']]).astype(int)
# Features and target
x = df1['stu_marks1'].values.reshape(-1, 1)
y = df1['stu_final_marks1']
# Train-test split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Train model
linear_regression1 = LinearRegression()
linear_regression1.fit(x_train, y_train)

# Predict
y_pred = linear_regression1.predict(x_test)

# Output results
print("Predictions:", y_pred)
print("MAE:", mean_absolute_error(y_test, y_pred))
print("R² (using model.score):", linear_regression1.score(x_test, y_test))  # ✅ CORRECT
print("R² (using r2_score):", r2_score(y_test, y_pred))                     # ✅ CORRECT
```
