```python
x=dataframe.iloc[:,:-1]
y=dataframe['Price']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
linear_reg=LinearRegression()
model1=Pipeline([
    ('linear_reg',linear_reg)
])

model={
    'Linear_regression':model1
}

for name,model in model.items():
    print(name)
    model.fit(x_train,y_train)
    y_housing_predication=model.predict(x_test)
    print(y_housing_predication)
    print(mean_absolute_error(y_test,y_housing_predication))
    print()
    print(r2_score(y_test,y_housing_predication))
```
