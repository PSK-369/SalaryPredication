```python
import seaborn as sns
from matplotlib import pyplot as plt
import numpy as np
import ssl
ssl._create_default_https_context = ssl._create_unverified_context
```

```python
print(sns.get_dataset_names())
```

```python
health_data=sns.load_dataset('penguins')
health_data
```

```python
health_data.isna().sum()
```

```python
health_data['bill_depth_mm ']=health_data['bill_depth_mm'].fillna(np.mean(health_data['bill_depth_mm']),axis=0)
```

```python
col1=['#ff4d01','#000000']
fonts={'size':20,'fontweight':'bold','color':'#ff4d01'}
#Count plot --> for catageious
plt.figure(figsize=(8,5))
sns.set_theme(style='white',palette=col1)
sns.countplot(data=health_data,x=health_data['sex'],width=0.45,legend='auto',stat='probability')
plt.xlabel('Gender wised',fontdict=fonts)
plt.ylabel('propability ratios',fontdict=fonts)
plt.title('Gender Wised Count plot',fontdict=fonts)
plt.savefig('countplot.png',dpi=1000)
plt.grid(True)
plt.show()
```

```python
col1=['#ff4d01','#ffffff']
fonts={'size':20,'fontweight':'bold','color':'#ff4d01'}
#Count plot --> for catageious
plt.figure(figsize=(10,10))
sns.set_theme(style='white',palette=col1)
sns.displot(data=health_data,x=health_data['sex'],y=health_data['island'])
plt.xlabel('Gender wised',fontdict=fonts)
plt.ylabel('Island ratios',fontdict=fonts)
plt.title('Gender Wised dis plot',fontdict=fonts)
plt.savefig('displot.png',dpi=1000)
plt.grid(True)
plt.show()
```

```python
col1=['#ff4d01','#ffffff']
fonts={'fontsize':20,'fontweight':'bold','color':'#ff4d01'}
#Count plot --> for catageious
plt.figure(figsize=(10,8))
sns.set_theme(style='white',palette=col1)
sns.barplot(data=health_data,x=health_data['sex'],y=health_data['island'],capsize=0.2,saturation=0.9,alpha=0.9)
plt.xlabel('Gender wised',fontdict=fonts)
plt.ylabel('Island ratios',fontdict=fonts)
plt.title('Gender Wised bar plot',fontdict=fonts)
plt.savefig('barplot.png',dpi=1000)
plt.grid(True)
plt.show()
```

```python
col1=['#ff4d01','#ffffff']
fonts={'size':20,'fontweight':'bold','color':'#ff4d01'}
#Count plot --> for catageious
plt.figure(figsize=(8,8))
sns.set_theme(style='white',palette=col1)
sns.violinplot(data=health_data,x=health_data['sex'],y=health_data['island'],orient='v')
plt.xlabel('Gender wised',fontdict=fonts)
plt.ylabel('Island ratios',fontdict=fonts)
plt.title('Gender Wised viloen plot',fontdict=fonts)
plt.savefig('viloen.png',dpi=1000)
plt.grid(True)
plt.show()
```
