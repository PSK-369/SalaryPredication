```python
import numpy as np 
import pandas as pd 
from matplotlib import pyplot as plt 
import seaborn as sns 
import statistics as sts 
import scipy.spatial.distance as  dts
import streamlit as st
import joblib as jlib
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,median_absolute_error
from sklearn.impute import SimpleImputer

from sklearn.datasets import make_regression
```

```python
dataFrame=pd.read_csv(r"C:\Shravan\PythonPratice\houseprice\Bangalore.csv",encoding_errors='ignore')

dataFrame.isna().sum()

x=dataFrame[['Area','Location','School','BED']]
y=dataFrame['Price']
```

```python
x
```

```python
labelen=LabelEncoder()
dataFrame['new_location_encoded']=labelen.fit_transform(dataFrame['Location'])
```

```python
x=dataFrame[['Area','new_location_encoded','School','BED']]
y=dataFrame['Price']
x
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

linear_reg=LinearRegression()
linear_reg.fit(x_train,y_train)
y_pred=linear_reg.predict(x_test)
print(y_pred)


```

```python
import joblib
joblib.dump(linear_reg,"house_predications.joblib")
```

```python
import streamlit as st
import pandas as pd
import joblib

model = joblib.load("house_predications.joblib")
st.title("Housing Predictions")

st.sidebar.write("Enter all required details")

with st.form(key='form_details'):
    with st.sidebar:
        Area = st.number_input('Area', min_value=1100, max_value=9999, value=3340)
        new_location = st.number_input('New Location', min_value=1, max_value=999, value=70)
        school = st.number_input('School', min_value=0, max_value=10, value=8)
        Bed = st.number_input('Bed', min_value=0, max_value=10, value=1)
        submit_button = st.form_submit_button(label='Submit')

if submit_button:
    data1 = {
        'area': Area,
        'newlocation': new_location,
        'school': school,
        'bed1': Bed
    }
    df = pd.DataFrame(data1, index=[0])
    prediction = model.predict(df)
    st.write(f"The prediction value is {prediction[0]:.2f}")
    st.balloons()
```
