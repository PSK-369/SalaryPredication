```python
import numpy as np
import pandas as pd 
from matplotlib import pyplot as plt 
import seaborn as sns 
from sklearn.preprocessing import StandardScaler,MinMaxScaler
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons

```

```python
x,y=make_moons(n_samples=1000,noise=0.25,random_state=42)
```

```python
cluster=pd.DataFrame(data=x,columns=['Feature1','Feature2'])
```

```python
cluster
```

```python
eps=0.2
minsamples=5
db=DBSCAN(eps=eps,min_samples=minsamples)
db.fit_predict(cluster)
```

```python
cl1=['#ff4d01','#ffffff']
sns.set_theme(style='white',palette=cl1)
plt.figure(figsize = (6, 5))
sns.scatterplot(x = 'Feature1', y = 'Feature2',data=cluster )
plt.title('DBSCAN Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```
