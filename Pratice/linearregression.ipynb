```python
#anomolies detection

class AnomoliesDetection:
    def __init__(self,data):
        self.data=data  
    def Check_function(self):
        mean=sts.mean(self.data)
        std=sts.stdev(self.data)
        cutoff = 3*std
        anomolies=[]
        postive_cutfoff=mean+cutoff
        negative_cutoff=mean-cutoff
        for each_anomiles in self.data:
            if each_anomiles > postive_cutfoff or each_anomiles < negative_cutoff:
                anomolies.append(each_anomiles)           
        return anomolies
       
anomiles_detection=AnomoliesDetection(df1['new_data'])
new_anomolies_data=anomiles_detection.Check_function()
print(new_anomolies_data)
```
