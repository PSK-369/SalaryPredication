```python
import numpy as np 
import pandas as pd 
from matplotlib import pyplot as plt 
import seaborn as sns 
import statistics as sts
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,PolynomialFeatures
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
```

```python
np.random.seed(42)
x=np.random.randint(low=1,high=200,size=(200)).astype(int)
y=np.random.uniform(low=1,high=200,size=(200))
data={
    'x_data':x,
    'y_data':y
}
```

```python
dataframe=pd.DataFrame(data,columns=['x_data','y_data'])
```

```python
dataframe['x_data'].sample(10)
```

```python
x=dataframe['x_data'].values.reshape(-1,1)
y=dataframe['y_data']
```

```python
class Anomiles_dection:
    def __init__(self,data):
        self.data=data
    def checking_function(self):
        anomiles=[]
        mean=sts.mean(self.data)
        std=sts.stdev(self.data)
        cutoff=3*std
        higher_bound=mean+(3*cutoff)
        lower_bound=mean-(3*cutoff)
        for each_anomile_point in self.data:
            if each_anomile_point > higher_bound or each_anomile_point < lower_bound:
                anomiles.append(each_anomile_point)
        return anomiles
    
anomiles_instance=Anomiles_dection(dataframe['x_data'])
results=anomiles_instance.checking_function()
print(f"result is - > {results}")
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
linear_reg=LinearRegression()
```

```python
model_poly1=PolynomialFeatures(degree=3,include_bias=False)
model_poly2=PolynomialFeatures(degree=4,include_bias=False)
model_poly3=Ridge(alpha=1)
model_poly4=Lasso(alpha=1)
model_poly5=ElasticNet(alpha=1)
```

```python
model1=Pipeline([
    ('model1',linear_reg),   
])
model3=Pipeline([
    ('model2',model_poly1),
    ('model3',linear_reg)   
])
model4=Pipeline([
      ('model4',model_poly3),
      ('model5',linear_reg),      
])

model2={
    'Linear_regression':model1,
    'Polynomial_regression':model3,
    #'model3':model4  
}


for name,model in model2.items():
    print(name)
    model.fit(x_train,y_train)
    y_pred=model.predict(x_test)
    print(y_pred)
```

```python
validation_score=cross_val_score(linear_reg,x,y,scoring='neg_mean_squared_error',cv=5)
mse_error=np.mean(validation_score)
print(abs(mse_error))
```
