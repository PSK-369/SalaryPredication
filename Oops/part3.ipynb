```python
from abc import ABC,abstractmethod

class BaseClass(ABC):
    @abstractmethod
    def addtion_function(self):
        pass 
    @abstractmethod
    def stringupper_function(self):
        pass 
    
class ChildClass(BaseClass):
    def addtion_function(self):
        a=10
        b=20 
        c=a+b
        return f"{c}"
    def stringupper_function(self):
        var1='shravan'
        return f"{var1.upper()}"
    

child_instance=ChildClass()
addtion=child_instance.addtion_function()
print(addtion)

string_upper=child_instance.stringupper_function()
print(string_upper)
```

```python
#polymorphosim

#Compile time -> to achive method overloading [same method name and different no of parameters]
#runtime polymorphism
class Compiletime_polymorphism:
    def addtion_function(self,var1,var2,var3=10):
        print('calling function is 2 var and 0 keyword')
        return f"{var1+var2+var3}"
    
class Runtime_polymorphism_parent:
    def printing_function(self,data1,data2,data3):
        return f"{data1+data2+data3}"
    
class Runtime_polymorphism_child(Runtime_polymorphism_parent):
    def printing_function(self,data1,data2,data3):
        return f"{data1+data2+data3}"
        
compile_instance=Compiletime_polymorphism()
compile1=compile_instance.addtion_function(var1=10,var2=20)
print(compile1)
 

```
