```python
class Father:
    def __init__(self,height,bone):
        self.height=height
        self.bone=bone

class Mother:
    def __init__(self,color,skintone):
        self.color=color
        self.skintone=skintone

class Son(Father,Mother):
    def __init__(self,height,bone,color,skintone):
        Father.__init__(self,height,bone)
        Mother.__init__(self,color,skintone)
    
    def Final_qualities(self):
        return f"{self.height} --> {self.bone} --> {self.color} --> {self.skintone}"


son_instance=Son('172.89cm','strong','yellow','brown')
result1=son_instance.Final_qualities()
print(result1)
```
